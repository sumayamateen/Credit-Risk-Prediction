## Age vs Default Analysis

```python
df = df.copy()

df['AGE'] = -df['DAYS_BIRTH'] / 365
df['AGE_GROUP'] = pd.cut(df['AGE'], bins=[0, 25, 35, 45, 55, 65, 75,100])
print(df[['DAYS_BIRTH', 'AGE', 'AGE_GROUP']])
```

```python
age_default = df.groupby("AGE_GROUP",  observed=False)["TARGET"].agg(["count", "mean"])
age_default["mean"] = age_default["mean"] * 100
age_default.columns = ["Number of People", "Default Rate %"]
print(age_default.round(2))
```

##  EMPLOYMENT vs  Default Analysis

```python
df['EMPLOYMENT_YEARS'] = df['DAYS_EMPLOYED'].copy()
df.loc[df['EMPLOYMENT_YEARS'] == 365243, 'EMPLOYMENT_YEARS'] = None
df['EMPLOYMENT_YEARS'] = -df['EMPLOYMENT_YEARS'] / 365

df['EMP_GROUP'] = pd.cut(df['EMPLOYMENT_YEARS'], bins=[0, 1, 3, 5, 10, 50],
                          labels=['<1 year', '1-3 years', '3-5 years', '5-10 years', '10+ years'])

emp_default = df.groupby('EMP_GROUP', observed=False)['TARGET'].agg(['count', 'mean'])
emp_default['mean'] = emp_default['mean'] * 100
emp_default.columns = ['Number of People', 'Default Rate %']
print(emp_default.round(2))
```

## Visualizations

```python
numeric_cols = ['TARGET', 'AMT_INCOME_TOTAL', 'AMT_CREDIT', 'CNT_CHILDREN', 
                'AGE', 'EXT_SOURCE_1', 'EXT_SOURCE_2', 'EXT_SOURCE_3']
corr = df[numeric_cols].corr()

plt.figure(figsize=(10, 8))
sns.heatmap(corr, annot=True, cmap='coolwarm', center=0, fmt='.2f')
plt.title('Correlation Between Key Features')
plt.tight_layout()
plt.show()
```

```python
def plot_default_rate(data, title, xlabel):
    plt.figure(figsize=(8, 5))
    data.plot(kind='bar', color='steelblue')
    plt.title(title)
    plt.xlabel(xlabel)
    plt.ylabel('Default Rate (%)')
    plt.xticks(rotation=45)
    plt.tight_layout()
    plt.show()

# Plot each
plot_default_rate(emp_default['Default Rate %'], 'Default Rate by Employment Duration', 'Years Employed')
plot_default_rate(housing_default['Default Rate %'], 'Default Rate by Homeownership', 'Owns Real Estate')
plot_default_rate(car_default['Default Rate %'], 'Default Rate by Car Ownership', 'Owns Car')
plot_default_rate(gender_default['Default Rate %'], 'Default Rate by Gender', 'Gender')
```
